# Результаты анализа стационарности временных рядов foF2

Проверка стационарности выполнена для дневных временных ходов foF2, сформированных из feature-строк, использованных в задаче 24-часового прогноза. В анализ включены станции, для которых был сформирован достаточный дневной ряд. Всего рассмотрено 35 станций с результатами ADF- и KPSS-критериев и 1 станция с недостаточным количеством наблюдений.

Для каждой станции рассчитывались ADF-критерий и KPSS-критерий. По сочетанию результатов станции были отнесены к трем группам: стационарные, тренд-стационарные и нестационарные. Из 35 станций 2 станции классифицированы как стационарные, 5 станций - как тренд-стационарные, 27 станций - как нестационарные. Одна станция в низких широтах имела недостаточное число дневных наблюдений для устойчивой проверки.

## Классификация станций

К стационарным рядам отнесены станции LL721 в низких широтах и EG931 в северных средних широтах. К тренд-стационарным рядам отнесены HO54K в южных средних широтах, EA653, FF051 и PA836 в северных средних широтах, а также THJ76 в северных высоких широтах. Остальные 27 станций были классифицированы как нестационарные.

| Класс ряда | Число станций |
|---|---:|
| Стационарный | 2 |
| Тренд-стационарный | 5 |
| Нестационарный | 27 |
| Недостаточно данных | 1 |

## Результаты ADF- и KPSS-критериев

Медианные значения p-value ADF-критерия составили: 0,243 для низких широт, 0,074 для южных средних широт, 0,016 для северных средних широт и 0,138 для северных высоких широт. Медианные значения p-value KPSS-критерия составили: 0,027 для низких широт, 0,010 для южных средних широт, 0,010 для северных средних широт и 0,010 для северных высоких широт.

Средние значения дневного foF2 по широтным зонам составили: 9,63 МГц для низких широт, 7,13 МГц для южных средних широт, 7,26 МГц для северных средних широт и 5,70 МГц для северных высоких широт. ## Итоговая сводка

По результатам ADF- и KPSS-критериев преобладают нестационарные временные ряды foF2: 27 из 35 станций с рассчитанными критериями были отнесены к нестационарным. Стационарные и тренд-стационарные ряды суммарно составили 7 станций. Максимальная доля стационарных и тренд-стационарных рядов получена в северных высоких широтах, однако эта группа включает только 2 станции.
